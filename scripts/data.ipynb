```python
import pandas as pd
import ast

df = pd.read_csv("../data/RAW_recipes.csv")

# Basic overview
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
print(df.describe())

# Check the string-list issue
print(type(df["steps"].iloc[0]))  # <class 'str'> — needs parsing!

# Outlier check on minutes
print(df["minutes"].quantile([0.5, 0.95, 0.99, 1.0]))
```

```python
print("sentinel minutes:", (df["minutes"] >= 2**31 - 1).sum())
print("minutes > 2880:", (df["minutes"] > 2880).sum())
print("zero-step recipes:", (df["n_steps"] == 0).sum())
print("duplicate ids:", df["id"].duplicated().sum())
```
